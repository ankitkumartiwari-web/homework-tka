```python
class Book:
    def __init__(self, price, name):
        self.p = price
        self.n = name

    def __add__(self, other):
        return f"Addition of price is {self.p + other.p} and concatenation of name is {self.n + other.n}"

    def __sub__(self, other):
        return f"Subtraction of price is {self.p - other.p} and strings do not support subtraction"

    def __mul__(self, other):
        return f"Multiplication of prices is {self.p * other.p} and strings can be multiplied by integers, not by other strings"

    def __truediv__(self, other):
        return f"Division of prices is {self.p / other.p} and strings do not support division"

    def __mod__(self, other):
        return f"Modulus of prices is {self.p % other.p} and strings do not support modulus"

    def __eq__(self, other):
        return f"Are prices same? {self.p == other.p}. Are the names equal? {self.n == other.n}"


catch1 = Book(5, "Book1")
catch2 = Book(2, "Book2")

print(catch1 + catch2)
print(catch1 - catch2)
print(catch1 * catch2)
print(catch1 / catch2)
print(catch1 % catch2)
```
