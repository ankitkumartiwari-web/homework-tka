```python
class Student:
    def __init__(self, name, roll_no, course, email):
        self.name = name
        self.roll_no = roll_no
        self.course = course
        self.email = email
    def display_student(self):
        print("-"*40)
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")
        print(f"Course: {self.course}")
        print(f"Email: {self.email}")
        print("-"*40)

student1 = Student("Rahul", "101", "Python Full Stack", "rahul@example.com")
student2 = Student("Anjali", "102", "Data Science", "anjali@example.com")
student1.display_student()
student2.display_student()
```

Task 3:
Scenario: Build a simple bank account system. 
Your Task: 
• Create a BankAccount class with account_number, holder_name and balance. 
• Initialize the values using __init__(). 
• Create deposit(amount) to add money. 
• Create withdraw(amount) to subtract money only when sufficient balance is available. 
• Create check_balance() to display the current balance. 
• Test deposit and withdrawal operations. 
Expected Output (example): 
Initial Balance: 10000 
Deposited: 3000 
Withdrawn: 2500 
Current Balance: 10500 
Bonus Challenge: Add a transaction_history list and store every deposit/withdrawal.

```python
class BankAccount:
    def __init__(self, account_number, account_holder, balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        amount = float(amount)
        if amount > 0:
            self.balance += amount
            print(f"Deposited: {amount} \nNew Balance: {self.balance}")

    def withdraw(self, amount):
        amount = float(amount)
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn: {amount} \nNew Balance: {self.balance}")
        else:
            print("Insufficient balance or invalid amount.")

    def check_balance(self):
        print(f"Account Holder: {self.account_holder} \nAccount Number: {self.account_number} \nBalance: {self.balance}")


catch = BankAccount("1234567890", "Rahul", 10000)
catch.deposit(3000)
catch.withdraw(2500)
catch.check_balance()
```

Scenario: An organization has different types of employees. Every employee has common information, while 
developers have additional information. 
Your Task: 
• Create a parent class Employee with name and salary. 
• Add a display_info() method to Employee. 
• Create a child class Developer that inherits Employee. 
• Add a programming_language attribute to Developer. 
• Create a method show_developer_info(). 
• Create a Developer object and display all information. 
Expected Output (example): 
Name: Neha 
Salary: 75000 
Programming Language: Python
Bonus Challenge: Add another child class Tester with a testing_tool attribute.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def display_info(self):
        print(f"Employee Name: {self.name} \nSalary: {self.salary}")

class Developer(Employee):
    def __init__(self, programming_language):
        self.programming_language = programming_language
    def show_developer_info(self):
        print(f"Programming Language: {self.programming_language}")
class Tester(Employee):
    def __init__(self, testing_tool):
        self.testing_tool = testing_tool
    def show_tester_info(self):
        print(f"Testing Tool: {self.testing_tool}")
dev1 = Developer("Python")
dev1.name = "Neha"
dev1.salary = 75000
dev1.display_info()
dev1.show_developer_info()

print("="*40)

test1 = Tester("Selenium")
test1.name = "Rohit"
test1.salary = 65000
test1.display_info()
test1.show_tester_info()
```

Your Task:

Expected Output (example):

```python
class Vehicle:
    def start(self):
        print("Vehicle started")
class Car(Vehicle):
    def start(self):
        print("Car starting with a key")
class Bike(Vehicle):
    def start(self):
        print("Bike is starting with a self-start button")
class Truck(Vehicle): 
    def start(self):
        print("Truck started with a heavy engine")
car = Car()
car.start()
bike = Bike()
bike.start()
truck = Truck()
truck.start()
vehicle = Vehicle()
vehicle.start()
```

Task 6:
Scenario: Build a small employee management system using OOP concepts. 
Your Task: 
• Create a parent class Employee with employee_id, name and salary. 
• Create Developer(Employee) with programming_language. 
• Create Tester(Employee) with testing_tool. 
• Use constructors in all classes. 
• Implement display_info() in the parent class. 
• Override display_info() in Developer and Tester to include role-specific information. 
• Create at least two developers and two testers. 
• Store all objects in a list and iterate through the list to display their details. 
Expected Output (example): 
The output should clearly identify each employee's ID, name, salary and role-specific 
information. 
Bonus Challenge: Add Manager(Employee) with a team_size attribute. Add a method to calculate annual salary.

```python
class Employee:
    def __init__(self, employee_id, name, salary):
        self.employee_id = employee_id
        self.name = name
        self.salary = salary

    def calculate_annual_salary(self):
        return self.salary * 12

    def get_role(self):
        return "Employee"

    def get_role_details(self):
        return "-"

    def display_info(self):
        print(
            f"| {self.employee_id:<11} "
            f"| {self.name:<12} "
            f"| {self.salary:<10.2f} "
            f"| {self.calculate_annual_salary():<13.2f} "
            f"| {self.get_role():<10} "
            f"| {self.get_role_details():<20} |"
        )


class Developer(Employee):
    def __init__(self, employee_id, name, salary, programming_language):
        super().__init__(employee_id, name, salary)
        self.programming_language = programming_language

    def get_role(self):
        return "Developer"

    def get_role_details(self):
        return self.programming_language


class Tester(Employee):
    def __init__(self, employee_id, name, salary, testing_tool):
        super().__init__(employee_id, name, salary)
        self.testing_tool = testing_tool

    def get_role(self):
        return "Tester"

    def get_role_details(self):
        return self.testing_tool


class Manager(Employee):
    def __init__(self, employee_id, name, salary, team_size):
        super().__init__(employee_id, name, salary)
        self.team_size = team_size

    def get_role(self):
        return "Manager"

    def get_role_details(self):
        return f"Team size: {self.team_size}"


employees = [
    Developer(101, "Rahul", 50000, "Python"),
    Developer(102, "Priya", 60000, "Java"),
    Tester(103, "Amit", 40000, "Selenium"),
    Tester(104, "Neha", 45000, "Postman"),
    Manager(105, "Raj", 80000, 8)
]

print("-" * 95)
print(
    f"| {'ID':<11} | {'Name':<12} | {'Salary':<10} "
    f"| {'Annual Salary':<13} | {'Role':<10} | {'Role Details':<20} |"
)
print("-" * 95)

for emp in employees:
    emp.display_info()

print("-" * 95)
```
