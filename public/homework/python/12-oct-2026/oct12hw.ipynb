Convert a given list into flatten list

```python
input_list = [1, [2, 3], [4, [5 , 6]], 7]

def flatten_list(nested_list):
    flat_list = [] # create an empty list to store the flattened elements
    for item in nested_list: # iterate through each item in the nested list
        if isinstance(item, list): # check if the item is a list
            flat_list.extend(flatten_list(item)) # recursively call the function on the sublist and extend the flat list with the result
        else:
            flat_list.append(item) # if the item is not a list, append it to the flat list
    return flat_list 

print(flatten_list(input_list))
```

```python
input_list = [1, [2, 3], [4, [5 , 6]], 7]

def ouput_list(lst):
    output_lst = [] # create an empty list to store the output

    for i in lst: # iterate through each element in the input list
        if type(i) == list: # check if the element is a list
            output_lst.extend(ouput_list(i)) # recursively call the function on the sublist and extend the output list with the result
        else:
            output_lst.append(i) # if the element is not a list, append it to the output list

    return output_lst

print(ouput_list(input_list))
```

```python
student_marks_db = {
    1: [78, 88, 69],
    2: [67, 89, 91],
    3: [77, 81, 92]
}
```

WAP to for average marks of each student and print the student id and average marks in a dictionary format

```python
for student_id, marks in student_marks_db.items():
    average_marks = sum(marks) / len(marks)
    print(f"Student ID: {student_id}, Average Marks: {average_marks:.2f}")
```

WAP to display topper roll_no and congratulate them

```python
lst = []

for student_id, marks in student_marks_db.items():
    lst.append(sum(marks))
topper = max(lst)
print(f"Congratualtions Roll Number {student_id} on scoring {topper} marks")
    
```

WAP to display subject wise topper

```python
subjects = ["Physics", "Chemistry", "Maths"]

for subject in range(len(subjects)):
    highest = 0

    for student_id, marks in student_marks_db.items():
        if marks[subject] > highest:
            highest = marks[subject]
            topper = student_id

    print(f"{subjects[subject]} Topper Marks: {highest}, Topper is Roll No: {topper}")
```
