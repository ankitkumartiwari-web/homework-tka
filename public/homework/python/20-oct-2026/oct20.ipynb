```python
# Class: Ankit
# Attributes:
#     name
#     age
#     college
#     skill
# Methods:
#     study()
#     code()

class Ankit:
    def __init__(self, name, age, college, skill):
        self.name = name
        self.age = age
        self.college = college
        self.skill = skill

    def study(self):
        print(f"{self.name} is studying {self.skill} in {self.college}.")

    def code(self):
        print(f"{self.name} is coding in Python using {self.skill} skills.")


# creating 5 objects of class Ankit
s1 = Ankit("Jay Sharma", 20, "MIT", "Python")
s2 = Ankit("Priya Verma", 21, "IIT Delhi", "Java")
s3 = Ankit("Rahul Singh", 19, "NIT Trichy", "C++")
s4 = Ankit("Sneha Patil", 22, "BITS Pilani", "Machine Learning")
s5 = Ankit("Karan Mehta", 23, "VIT", "Web Development")

students = [s1, s2, s3, s4, s5]

# displaying the attributes of every object
for i in students:
    print("-"*40)
    print(f"Name: {i.name}")
    print(f"Age: {i.age}")
    print(f"College: {i.college}")
    print(f"Skill: {i.skill}")

# calling the methods on the objects
for i in students:
    i.study()

print("="*40)

for i in students:
    i.code()
```

```python
# Class: Bulb
# Attributes:
#     color
#     price
#     power
#     is_on
# Methods:
#     turn_on()
#     turn_off()
#     show_details()

class Bulb:
    def __init__(self, color, price, power):
        self.color = color
        self.price = price
        self.power = power
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        print(f"The {self.color} bulb ({self.power}W) is switched ON.")

    def turn_off(self):
        self.is_on = False
        print(f"The {self.color} bulb ({self.power}W) is switched OFF.")

    def show_details(self):
        print("-"*40)
        print(f"Color: {self.color}")
        print(f"Price: {self.price}")
        print(f"Power: {self.power}W")
        print(f"State: {'ON' if self.is_on else 'OFF'}")


# creating 5 objects of class Bulb
b1 = Bulb("White", 100, 60)
b2 = Bulb("Yellow", 80, 40)
b3 = Bulb("Blue", 150, 9)
b4 = Bulb("Red", 120, 100)
b5 = Bulb("Green", 90, 15)

bulbs = [b1, b2, b3, b4, b5]

# displaying the attributes of every object
for i in bulbs:
    i.show_details()

# switching on every bulb
for i in bulbs:
    i.turn_on()

print("="*40)

# switching off every bulb
for i in bulbs:
    i.turn_off()

print("="*40)

# checking the state of every bulb
for i in bulbs:
    i.show_details()
```

```python
# Class: Keyboard
# Attributes:
#     mechanical
#     color
#     price
#     layout
# Methods:
#     type()
#     press_key()
#     show_details()

class Keyboard:
    def __init__(self, mechanical, color, price, layout):
        self.mechanical = mechanical
        self.color = color
        self.price = price
        self.layout = layout

    def type(self, text="Hello World"):
        print(f"Typing '{text}' on the {self.color} keyboard.")

    def press_key(self, key="Enter"):
        print(f"The key '{key}' is pressed on the {self.color} keyboard.")

    def show_details(self):
        print("-"*40)
        print(f"Type: {'Mechanical' if self.mechanical else 'Membrane'}")
        print(f"Color: {self.color}")
        print(f"Price: {self.price}")
        print(f"Layout: {self.layout}")


# creating 5 objects of class Keyboard
k1 = Keyboard(True, "Black", 4500, "US Full Size")
k2 = Keyboard(False, "White", 1200, "Indian Compact")
k3 = Keyboard(True, "RGB Backlit", 8500, "UK Full Size")
k4 = Keyboard(False, "Grey", 2000, "Indian Full Size")
k5 = Keyboard(True, "Blue", 6000, "US Tenkeyless")

keyboards = [k1, k2, k3, k4, k5]

# displaying the attributes of every object
for i in keyboards:
    i.show_details()

# typing and pressing a key on every keyboard
for i in keyboards:
    i.type("Hello Python")
    i.press_key("Enter")
    print("-"*40)

# showing the details of every keyboard
for i in keyboards:
    i.show_details()
```

```python
# Class: UNO
# Attributes:
#     color
#     number
#     symbol
#     player
# Methods:
#     draw_card()
#     play_card()
#     show_details()

class UNO:
    def __init__(self, color, number, symbol, player):
        self.color = color
        self.number = number
        self.symbol = symbol
        self.player = player

    def draw_card(self):
        print(f"{self.player} draws the {self.color} card from the deck.")

    def play_card(self):
        card = self.number if self.number != "None" else self.symbol
        print(f"{self.player} plays the {self.color} {card} card on the pile.")

    def show_details(self):
        print("-"*40)
        print(f"Color: {self.color}")
        print(f"Number: {self.number}")
        print(f"Symbol: {self.symbol}")
        print(f"Player: {self.player}")


# creating 5 objects of class UNO
c1 = UNO("Red", "7", "None", "Aarav")
c2 = UNO("Blue", "5", "None", "Priya")
c3 = UNO("Green", "0", "None", "Rahul")
c4 = UNO("Yellow", "None", "Skip", "Sneha")
c5 = UNO("Black", "None", "Wild", "Karan")

cards = [c1, c2, c3, c4, c5]

# displaying the attributes of every object
for i in cards:
    i.show_details()

# every player draws a card
for i in cards:
    i.draw_card()

print("="*40)

# every player plays a card
for i in cards:
    i.play_card()
```

```python
# Class: Maggie
# Attributes:
#     flavor
#     price
#     weight
# Methods:
#     cook()
#     eat()
#     show_details()

class Maggie:
    def __init__(self, flavor, price, weight):
        self.flavor = flavor
        self.price = price
        self.weight = weight

    def cook(self):
        print(f"Cooking the {self.flavor} Maggi ({self.weight}g pack) for 2 minutes.")

    def eat(self):
        print(f"Yum! The {self.flavor} Maggi is eaten.")

    def show_details(self):
        print("-"*40)
        print(f"Flavor: {self.flavor}")
        print(f"Price: {self.price}")
        print(f"Weight: {self.weight}g")


# creating 5 objects of class Maggie
m1 = Maggie("Masala", 12, 70)
m2 = Maggie("Hot and Sour", 14, 75)
m3 = Maggie("Tomato", 13, 80)
m4 = Maggie("Chicken", 15, 99)
m5 = Maggie("Peri Peri", 16, 105)

maggies = [m1, m2, m3, m4, m5]

# displaying the attributes of every object
for i in maggies:
    i.show_details()

# cooking every pack
for i in maggies:
    i.cook()

print("="*40)

# eating every pack
for i in maggies:
    i.eat()
```
