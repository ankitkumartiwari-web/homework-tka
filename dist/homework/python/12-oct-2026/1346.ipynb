Highest Score of each student

```python
db = {
    1: [78, 97, 86],
    2: [99, 94, 92],
    3: [78, 90, 98]
}

subject = ["Physics", "Chemistry", "Maths"]

lst = []
n = 1
for i,j,k in db.values(): # this will unpack the values of each roll number into i, j, k
    topper = max(i,j,k) # maximum marks in each roll number using max function
    lst.append(topper) # appending the topper marks in the list

#print(lst) #[97, 99, 98]
for mark, sub in zip(lst, subject): # zipping the topper marks with the subject names
    print(f"Roll Number {n} is topper in {sub} with marks {mark}")
    n += 1 # incrementing the roll number for each iteration
```

Subject Wise Topper

```python
db = {
    1: [78, 97, 86],
    2: [99, 94, 92],
    3: [78, 90, 98]
}

subject = ["Physics", "Chemistry", "Maths"]
lst = []

for i in range(len(subject)): # 0, 1, 2
    highest = 0 # Initialize the highest mark to 0
    roll = 0 # Initialize the roll number of the topper to 0

    for roll_number, marks in db.items(): # Iterate through the database
        if marks[i] > highest: # If the current mark is higher than the highest mark found so far
            highest = marks[i] # Update the highest mark
            roll = roll_number # Update the roll number of the topper

    lst.append([roll, highest]) # Append a list of (roll number, highest mark) to the list

for sub, topper in zip(subject, lst): # Iterate through the subjects and the list of toppers using zip
    print(f"Roll Number {topper[0]} is topper in {sub} with marks {topper[1]}")


# zip() function is used to combine two or more iterables (like lists, tuples, etc.) into a single iterable. It pairs elements from each iterable based on their position. The first elements of each iterable are paired together, then the second elements, and so on. If the iterables are of different lengths, zip stops creating pairs when the shortest iterable is exhausted.
```
